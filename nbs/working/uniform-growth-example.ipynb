```python
from cd.algs.cd_poly import CDPolynomial
import numpy as np
import matplotlib.pyplot as plt
```

```python
m = 1000
x = 2 * np.random.rand(m)[:, np.newaxis] - 1
```

```python
x0 = np.array([[0.99]])
x1 = np.array([[1.1]])

vals_x0 = []
vals_x1 = []

ds = [i for i in range(1, 11)]

for d in ds:
    p = CDPolynomial(x, degree=d)
    vals_x0.append(p(x0)[0])
    vals_x1.append(p(x1)[0])

plt.plot(ds, vals_x0, marker='x', label='x=0.99')
plt.plot(ds, vals_x1, marker='s', label='x=1.1')
plt.xlabel('degree d')
plt.ylabel('CD polynomial')
plt.yscale('log')
plt.legend()
```
