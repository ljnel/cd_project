```python
import pandas as pd
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
```

```python
df_sa = pd.read_json('sa.json', lines=True)
df_sa['no_fail'] = df_sa['no_fail'].fillna(1)
cols = {'t': 'Time',
        'no_fail': 'No Fail',
        'weight_acc': 'Weighted Acc.',
        'edr': 'EDR',
        'ttd': 'TTD',
        'env': 'Environment',
        'method': 'Method'
}
df_sa = df_sa.rename(columns=cols)
df_sa = df_sa.replace({
    'cd': 'CD',
    'cd_ae': 'CD-AE',
    'dist': 'Dist',
    'ae_rec': 'AE-REC',
    'cd_rp': 'CD-RP'
})
df_sa = df_sa.replace({
    'inv_pend': 'Inv. Pend.',
    'hopper': 'Hopper',
    'half_cheetah': 'Half Cheetah',
    'ant': 'Ant',
    'humanoid': 'Humanoid'
})

#drop_rows = [0, 1, 8, 9, 10, 12, 13, 14, 16, 17, 18, 23]
df_sa
```

```python
df_s = pd.read_json('s.json', lines=True)
df_s['no_fail'] = df_s['no_fail'].fillna(1)
cols = {'t': 'Time',
        'no_fail': 'No Fail',
        'weight_acc': 'Weighted Acc.',
        'edr': 'EDR',
        'ttd': 'TTD',
        'env': 'Environment',
        'method': 'Method'
}
df_s = df_s.rename(columns=cols)
df_s = df_s.replace({
    'cd': 'CD',
    'cd_ae': 'CD-AE',
    'dist': 'Dist',
    'ae_rec': 'AE-REC',
    'cd_rp': 'CD-RP'
})
df_s = df_s.replace({
    'inv_pend': 'Inv. Pend.',
    'hopper': 'Hopper',
    'half_cheetah': 'Half Cheetah',
    'ant': 'Ant',
    'humanoid': 'Humanoid'
})

#drop_rows = [0, 1, 8, 9, 10, 12, 13, 14, 16, 17, 18, 23]
df_s
```

```python
fig, ax = plt.subplots(ncols=3, figsize=(12, 4))

ax[0].plot(df_sa['Weighted Acc.'], '-x', color='red', label='SA')
ax[0].plot(df_s['Weighted Acc.'], '-x', color='blue', label='S')
ax[0].set_title('Weighted Acc.')
ax[0].legend()

ax[1].plot(df_sa['EDR'], '-x', color='red', label='SA')
ax[1].plot(df_s['EDR'], '-x', color='blue', label='S')
ax[1].set_title('EDR')
ax[1].legend()

ax[2].plot(df_sa['TTD'], '-x', color='red', label='SA')
ax[2].plot(df_s['TTD'], '-x', color='blue', label='S')
ax[2].set_title('TTD')
ax[2].legend()
```
