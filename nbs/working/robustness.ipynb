```python
import numpy as np
import matplotlib.pyplot as plt
from cd.algs.cd_poly import CDPolynomial
```

```python
x = np.random.randn(300, 2)
```

```python
plt.scatter(*x.T)

p = CDPolynomial(x, degree=5)
pt = [[4., 4.]]
q = CDPolynomial(np.concatenate((x, pt)), degree=5, eps=1e1, method='chol')
plt.scatter(pt[0][0], pt[0][1], color='red')

plt.xlim(-5, 5)
plt.ylim(-5, 5)
p.plot(plt.gca(), colors='green')
q.plot(plt.gca(), colors='red')
```

```python
n = 10000
x = np.sort(np.random.rand(n))[:, np.newaxis]
p = CDPolynomial(x, degree=3, verbose=True)
plt.scatter(x, np.zeros(n), color='red', s=2, marker='x')
t = np.linspace(-0.1, 1.1, 10000)[:, np.newaxis]
plt.plot(t, np.exp(-p(t)))
```
