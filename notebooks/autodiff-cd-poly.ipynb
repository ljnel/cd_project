```python
from algs.cd_poly import CDPolynomial
import jax
from jax import random, grad
import jax.numpy as jnp

import matplotlib.pyplot as plt
```

```python
key = random.key(0)
z = random.normal(key, (1000, 2))
```

```python
p = CDPolynomial(z, degree=3)
```

```python
z_test = -1*jnp.ones(2)[None, :]
grad(lambda x: p(x)[0])(z_test)
```

```python
plt.scatter(*z.T)
p.plot(plt.gca())
```
